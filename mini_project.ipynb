1. ***Import*** **libraries**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

2. ***Load the Dataset***

```python
df = pd.read_csv("netflix_titles.csv.zip")
print(df.head())
```

```python
print(df.tail())
```

3. ***Understand the Dataset***

```python
print(df.shape)
print(df.columns)
print(df.info())
```

```python
print("Total Netflix titles: ", len(df))
```

4. ***Check for Missing Values***

```python
print(df.isnull().sum())
```

```python
missing = df.isnull().sum()
print(missing[missing > 0])
```

```python
sns.heatmap(df.isnull(), cbar=False)
plt.title("Where is netflix data missing?")
plt.show()
```

5. ***Movies vs TV shows***

```python
print(df["type"].value_counts)
```

```python
sns.countplot(data=df, x="type")
plt.title("Netflix Movies vs TV shows")
plt.xlabel("Type")
plt.ylabel("Nuber of titles")
plt.show()
```

6. ***Which year has the most netflix releases***

```python
year_count = df["release_year"].value_counts().sort_index()
print(year_count)
```

```python
plt.figure(figsize=(12,5))
plt.plot(year_count.index,year_count.values)
plt.title("Netflix releases over the years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```
